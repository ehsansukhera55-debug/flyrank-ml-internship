# 1. Data Contract — CTR / Engagement Opportunity Scoring

## 1.1 Selected Lane

CTR / Engagement Opportunity Scoring.

## 1.2 Business Problem

Identify website content pages that receive search impressions but generate fewer clicks than expected for their search ranking positions. The goal is to prioritize pages for human review.

### 1.3 Data Grain — Verified

Each row represents one content page belonging to one client on one reporting date.

**Confirmed unique key:**
- `report_date`
- `client_hash_id`
- `content_hash_id`

Verification on March 2026 data:
- Total rows: 9,841,378
- Unique key combinations: 9,841,378
- Duplicate rows: 0

Therefore, the proposed data grain is confirmed.

### 1.4 Data Quality and Availability Verification

The March 2026 dataset contains 31 distinct reporting dates, from March 1 to March 31, 2026.

Google Search Console availability:
- Total rows: 9,841,378
- GSC available rows: 3,611,061 (36.7%)
- Excluded rows: 6,230,317 (63.3%)

Only records where `gsc_data_available IS TRUE` will be included in the primary CTR analysis. Records with unavailable GSC data will not be interpreted as zero clicks or zero impressions.
## 1.5 Development Time Window

March 1–31, 2026.

The June 2026 sample will not be used for feature development or label design.

## 1.6 Prediction Target

Proposed target: Identify content pages whose future search CTR underperforms the expected CTR for comparable search ranking positions.

The exact future observation window and label threshold will be defined after inspecting the available data.

## 1.7 Deliberate Exclusion

Records where `gsc_data_available IS NOT TRUE` will be excluded from the primary CTR analysis because reliable Search Console performance data is unavailable.

Missing GA4 values will not be treated as zero engagement.## 1.8 Missing-Value Analysis

The March 2026 dataset contains no missing values in:
- report_date
- client_hash_id
- content_hash_id
- gsc_impressions
- gsc_clicks

There are 6,230,317 missing values in gsc_avg_position.
This count matches the number of records where GSC data is unavailable.

Decision:
- Filter records using gsc_data_available IS TRUE.
- Do not replace missing average positions with zero.
- Do not treat unavailable GSC data as zero clicks or zero impressions.

```python

grain_query = f"""
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT (
        report_date,
        client_hash_id,
        content_hash_id
    )) AS unique_combinations,
    COUNT(*) - COUNT(DISTINCT (
        report_date,
        client_hash_id,
        content_hash_id
    )) AS duplicate_rows
FROM read_parquet('{daily_path}')
"""

grain_result = con.sql(grain_query).df()
display(grain_result)
```

```python
coverage_query = f"""
SELECT
    COUNT(*) AS total_rows,
    MIN(report_date) AS first_date,
    MAX(report_date) AS last_date,
    COUNT(DISTINCT report_date) AS distinct_dates
FROM read_parquet('{daily_path}')
"""

coverage_result = con.sql(coverage_query).df()
display(coverage_result)
```

```python
availability_query = f"""
SELECT
    COUNT(*) AS total_rows,

    COUNT(*) FILTER (
        WHERE gsc_data_available IS TRUE
    ) AS gsc_available_rows,

    COUNT(*) FILTER (
        WHERE gsc_data_available IS NOT TRUE
    ) AS excluded_rows

FROM read_parquet('{daily_path}')
"""

availability_result = con.sql(availability_query).df()
display(availability_result)
```

```python
null_check_query = f"""
SELECT
    COUNT(*) AS total_rows,

    COUNT(*) FILTER (
        WHERE report_date IS NULL
    ) AS null_report_date,

    COUNT(*) FILTER (
        WHERE client_hash_id IS NULL
    ) AS null_client_id,

    COUNT(*) FILTER (
        WHERE content_hash_id IS NULL
    ) AS null_content_id,

    COUNT(*) FILTER (
        WHERE gsc_impressions IS NULL
    ) AS null_impressions,

    COUNT(*) FILTER (
        WHERE gsc_clicks IS NULL
    ) AS null_clicks,

    COUNT(*) FILTER (
        WHERE gsc_avg_position IS NULL
    ) AS null_avg_position

FROM read_parquet('{daily_path}')
"""

null_result = con.sql(null_check_query).df()
display(null_result)
```

```python
distribution_query = f"""
SELECT
    COUNT(*) AS available_rows,

    MIN(gsc_impressions) AS min_impressions,
    APPROX_QUANTILE(gsc_impressions, 0.50) AS median_impressions,
    APPROX_QUANTILE(gsc_impressions, 0.90) AS p90_impressions,
    MAX(gsc_impressions) AS max_impressions,

    MIN(gsc_clicks) AS min_clicks,
    APPROX_QUANTILE(gsc_clicks, 0.50) AS median_clicks,
    MAX(gsc_clicks) AS max_clicks,

    MIN(gsc_avg_position) AS min_position,
    APPROX_QUANTILE(gsc_avg_position, 0.50) AS median_position,
    APPROX_QUANTILE(gsc_avg_position, 0.90) AS p90_position,
    MAX(gsc_avg_position) AS max_position

FROM read_parquet('{daily_path}')
WHERE gsc_data_available IS TRUE
"""

distribution_result = con.sql(distribution_query).df()
display(distribution_result)
```
